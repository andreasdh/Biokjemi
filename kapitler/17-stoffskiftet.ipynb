## Hva er stoffskifte? {#sec-hva-er-stoffskifte}

**Stoffskifte** (metabolisme) er hele nettverket av kjemiske reaksjoner i en levende celle. Det består av to deler som går hånd i hånd:

- **Katabolisme:** å bryte ned molekyler (for eksempel glukose) og frigjøre energi.
- **Anabolisme:** å bygge opp molekyler (for eksempel proteiner og fett), noe som koster energi.

Mellomproduktene og produktene kalles **metabolitter**. Det er som regel små molekyler, og de henger sammen i **metabolske veier**: kjeder av reaksjoner der produktet fra ett trinn er substratet i neste.

Cellen trenger energi til tre hovedformål: **mekanisk arbeid** (muskelsammentrekning, transport av vesikler), **transport av molekyler** mot en konsentrasjonsgradient og **syntese** av nye molekyler. Vi mennesker er *kjemotrofer*: vi får energien fra å oksidere maten vi spiser. Planter er *fototrofer*: de bruker sollys, og lager først sukker ved fotosyntese.

{{< include ../_figurer/stoffskifte-trinn.html >}}

::: {.callout-tip}
## Hvorfor mange små trinn?
Hvis cellen brente glukose i ett eneste trinn, ville nesten all energien gått tapt som varme. Ved å bryte ned glukose i mange små trinn, hvert med sitt enzym, fanger cellen energien i håndterbare porsjoner (ATP og NADH). Det er som å gå ned en trapp og samle opp mynter på hvert trinn, i stedet for å hoppe ned fra taket.
:::

Metabolske veier er **sterkt regulert**. Cellen gjør det på tre måter: ved å endre *mengden* av enzymer, ved å endre enzymenes *aktivitet* (for eksempel med aktivatorer og hemmere) og gjennom *tilgangen på substrater*. I neste kapittel ser du hvordan det virker i praksis.

## Energi og konsentrasjoner {#sec-energi-konsentrasjoner}

Fra kapittel 2 husker du at **ΔG** (fri energi) avgjør om en reaksjon går av seg selv: ΔG < 0 betyr at reaksjonen kan gå. Du har sikkert også sett en **ΔG°′** i tabeller. De to er ikke det samme.

- **ΔG°′** er endringen i fri energi under *standardbetingelser*: alle stoffer ved 1 M, pH 7. Den er en fast egenskap ved reaksjonen, og forteller hvor *likevekten* ligger.
- **ΔG** er endringen i fri energi under de *faktiske* betingelsene i cellen. Den forteller hvilken vei reaksjonen går akkurat nå, og hvor mye «driv» den har.

Sammenhengen er

$$\Delta G = \Delta G^{\circ\prime} + RT \ln Q, \qquad Q = \frac{[\text{produkter}]}{[\text{substrater}]}$$

Her er Q forholdet mellom produkt- og substratkonsentrasjonene akkurat nå (med støkiometriske eksponenter), R er gasskonstanten og T temperaturen (RT ≈ 2,58 kJ/mol ved 37 °C). Når Q er lik likevektskonstanten K, er ΔG = 0, og reaksjonen står stille.

Prøv selv. Dra i Q, og se hva som skjer med retningen:

{{< include ../_interaktiv/gl-dg.qmd >}}

Tre ting å legge merke til:

1. **En reaksjon går alltid mot likevekt.** ΔG viser hvor langt unna likevekten den er. Jo mer negativ, jo større driv.
2. **Konsentrasjonene kan snu «ugunstige» reaksjoner.** Selv med ΔG°′ = +24 kJ/mol kan ΔG bli nær null hvis produktene fjernes raskt nok, slik at Q blir svært liten. Det er akkurat det som skjer i aldolase-trinnet i glykolysen.
3. **Cellen er ikke i likevekt.** Den holder konsentrasjoner av ATP, ADP og mellomprodukter langt fra likevekt, og det er det som gjør at stoffene *strømmer*. En celle i likevekt (ΔG = 0 overalt) er en død celle.

### Koblede reaksjoner

En reaksjon som ikke går av seg selv, kan kobles til en som gjør det. Hvis to reaksjoner har et felles mellomprodukt, legger vi bare sammen ΔG°′:

| Reaksjon | ΔG°′ (kJ/mol) |
|:---|---:|
| A ⇌ B + C | +21 |
| B ⇌ D | −34 |
| **A ⇌ C + D (sum)** | **−13** |

Den første reaksjonen er ugunstig alene, men summen er gunstig. Slik drives mange biokjemiske reaksjoner. Den viktigste «gunstige partneren» er hydrolysen av ATP.

## ATP: cellens energivaluta {#sec-atp}

**ATP** (adenosintrifosfat) består av adenin, ribose og tre fosfatgrupper. Når den ytterste fosfatgruppen spaltes av med vann, dannes ADP og uorganisk fosfat (Pᵢ):

{{< include ../_figurer/atp.html >}}

$$\text{ATP} + \text{H}_2\text{O} \rightleftharpoons \text{ADP} + \text{P}_i \qquad \Delta G^{\circ\prime} = -30{,}5 \text{ kJ/mol}$$

Reaksjonen er svært gunstig. Spaltes ATP i stedet til AMP og pyrofosfat (PPᵢ), blir ΔG°′ = −45,6 kJ/mol, og pyrofosfatet hydrolyseres videre. Slike gunstige reaksjoner bruker cellen til å drive ugunstige.

### Hvorfor gir ATP-hydrolyse så mye energi?

Det er ikke «i» en spesiell binding energien sitter. Det er forskjellen mellom ATP og produktene som teller. Fire grunner, og de passer fint med de tre pilarene fra kapittel 2:

::: {.callout-note}
## Fire grunner til at ATP-hydrolyse er gunstig
1. **Frastøtning (elektrontetthet og vekselvirkninger).** ATP har omtrent fire negative ladninger tett sammen, og de frastøter hverandre. Spaltes et fosfat av, avtar frastøtningen.
2. **Entropi (termodynamikk).** Ett molekyl blir til to (ADP og Pᵢ), som gir flere mulige fordelinger av energi og partikler.
3. **Resonansstabilisering (elektrontetthet).** Fritt fosfat har flere resonansformer enn fosfat bundet i ATP. Elektrontettheten kan fordeles over flere oksygener, som stabiliserer produktet.
4. **Hydrering (vekselvirkninger).** ADP og Pᵢ er bedre stabilisert av vann (ion–dipol og hydrogenbindinger) enn ATP.
:::

ATP er altså *termodynamisk ustabil* (reaksjonen med vann er gunstig), men *kinetisk stabil*: den reagerer ikke med vann av seg selv, for aktiveringsenergien er høy. Det er enzymene som bestemmer når ATP skal spaltes, og det gjør ATP perfekt for regulering.

### ATP i cellen er enda mer energirik

ΔG°′ = −30,5 kJ/mol gjelder under standardbetingelser. I cellen holder stoffskiftet forholdet [ATP]/[ADP] høyt, langt fra likevekt. Da blir Q lite, og ΔG blir mye mer negativ. Med [ATP] = 1,85 mM, [ADP] = 0,14 mM og [Pᵢ] = 1,0 mM får vi

$$\Delta G = -30{,}5 + RT \ln \frac{[\text{ADP}][\text{P}_i]}{[\text{ATP}]} \approx -30{,}5 + 2{,}58 \cdot \ln(7{,}6\cdot 10^{-5}) \approx -55 \text{ kJ/mol}$$

Altså nesten dobbelt så mye energi per ATP som tabellverdien. Cellen *holder ATP høyt*, og det gir den en kraftig drivkraft.

### ATP må hele tiden lages på nytt

Kroppen har lite ATP, men enormt forbruk. ATP gir energi til bevegelse, aktiv transport, biosyntese og signaloverføring, og blir til ADP. Forbrenning av brensel (eller fotosyntese hos planter) sørger for at ADP blir til ATP igjen. En av hovedoppgavene til katabolismen er derfor å **regenerere ATP**.

### Fosfatstigen: hvem kan gi fosfat til hvem?

Ikke alle fosfatforbindelser er like energirike. For hver forbindelse kan vi måle ΔG°′ for hydrolyse av fosfatet. Jo mer negativ, jo bedre fosfatgiver. Plasserer vi dem etter størrelse, får vi en «stige»: fosfat kan bare gå *nedover* stigen av seg selv. ATP står i midten, så den kan både gi fra seg fosfat (til sukker) og få fosfat (fra mer energirike giver som PEP).

Prøv et par eksempler. Disse tre kommer du til å møte igjen i glykolysen:

{{< include ../_interaktiv/gl-fosfat.qmd >}}

Overføringen er egentlig bare to koblede reaksjoner: hydrolyse av giveren (gunstig) og dannelsen av mottakerens fosfat (ugunstig). ΔG°′ for overføringen er differansen. Slik får cellen ATP til å fosforylere glukose, og slik får den PEP til å lage ATP.

### Andre aktiverte bærere

ATP er den vanligste bæreren, men ikke den eneste. Cellen har flere «bærere» som hver frakter sin type gruppe:

| Bærer | Gruppen som bæres | Vitaminforløper |
|:---|:---|:---|
| ATP | fosforyl (fosfat) | |
| NADH og NADPH | elektroner | niacin (B₃) |
| FADH₂ | elektroner | riboflavin (B₂) |
| Koenzym A | acyl (for eksempel acetyl) | pantotenat (B₅) |

## Elektroner: oksidasjon gir energi {#sec-elektroner}

Fra katabolismen er hovedtanken enkel: **oksidasjon av brensel driver dannelsen av ATP**. Hvorfor frigjør oksidasjon energi? Svaret ligger i elektrontettheten.

Se på karbon. I metan (CH₄) er karbonet bundet til fire hydrogen, som er litt mindre elektronegative, så karbonet har mye elektrontetthet. Jo flere bindinger til oksygen (som trekker elektronene bort), jo mindre elektrontetthet er det igjen på karbonet. Vi sier at karbonet **oksideres**. Trykk deg gjennom trappen:

{{< include ../_interaktiv/gl-oksidasjon.qmd >}}

Regelen er: **jo mer redusert et stoff er, jo mer energi frigjøres når det oksideres.** Det forklarer at fettsyrer (med mange C–H-bindinger) gir mer energi per gram enn karbohydrater (som allerede er delvis oksidert, med mange C–O-bindinger). Men fett passer ikke for alle celler: hjernen og de røde blodcellene er avhengige av glukose.

### Elektronbærerne

Cellen bruker ikke oksygen direkte på glukose. Den overfører elektronene til *bærere* som frakter dem videre til steder der energien kan utnyttes. Den viktigste er **NAD⁺** (nikotinamid-adenin-dinukleotid):

- **NAD⁺** er den oksiderte formen og tar imot elektroner (to elektroner og ett proton, som et *hydridion*, H⁻). Da blir den til **NADH**.
- **NADH** leverer elektronene videre, hovedsakelig til elektrontransportkjeden, som bruker dem til å lage ATP.
- **FAD** virker på samme måte (FAD → FADH₂).
- **NADPH** er en nær slektning som brukes til biosyntese. Den leverer elektroner når cellen skal *bygge opp* molekyler (reduserende kraft).

Dehydrogenering (fjerning av H fra en alkohol) skjer slik, og reaksjonen er en av de vanligste i stoffskiftet:

{{< include ../_interaktiv/gl-nad.qmd >}}

$$\text{R–CHOH–R}' + \text{NAD}^+ \rightleftharpoons \text{R–CO–R}' + \text{NADH} + \text{H}^+$$

::: {.callout-note}
## To valutaer
Cellen har to «valutaer»: **fosfat** (ATP) og **elektroner** (NADH og NADPH). ATP driver arbeid direkte. Elektronene i NADH og FADH₂ må gjennom elektrontransportkjeden for å bli til ATP, som du ser i et senere kapittel.
:::

## Oppsummering {#sec-stoffskifte-oppsummering}

::: {.nokkelpunkt}
**Stoffskifte** er summen av katabolisme (nedbrytning, frigjør energi) og anabolisme (oppbygging, krever energi).

**ΔG = ΔG°′ + RT ln Q.** ΔG°′ er standardverdien og viser hvor likevekten ligger. ΔG avhenger av konsentrasjonene og avgjør retningen. Cellen holder seg langt fra likevekt.

**ATP** er energivalutaen: hydrolysen er gunstig fordi frastøtningen avtar, entropien øker, produktene er resonansstabilisert og bedre hydrert. ATP er kinetisk stabil og kan reguleres av enzymer. Fosfatstigen viser hvem som kan gi fosfat til hvem.

**Oksidasjon** frigjør energi, og elektronene fraktes av NADH, FADH₂ og NADPH.
:::

## Oppgaver {#sec-stoffskifte-oppgaver}

**Oppgave 1.** For en reaksjon A ⇌ B er ΔG°′ = +10 kJ/mol ved 37 °C (RT = 2,58 kJ/mol). Hvilken verdi må Q ha for at reaksjonen skal stå i likevekt? Hva skjer hvis Q er mindre enn dette?

::: {.callout-tip collapse="true"}
## Svarforslag
I likevekt er ΔG = 0 og Q = K = e^(−ΔG°′/RT) = e^(−10/2,58) ≈ 0,02. Det betyr [B]/[A] ≈ 0,02. Hvis Q er mindre (mindre B enn likevekten tilsier), blir ΔG negativ og reaksjonen går forover, selv om ΔG°′ er positiv.
:::

**Oppgave 2.** Forklar hvorfor en celle der alle reaksjoner var i likevekt, ikke ville være i live.

::: {.callout-tip collapse="true"}
## Svarforslag
I likevekt er ΔG = 0 for alle reaksjoner, så det er ingen netto flyt og ingen mulighet til å utføre arbeid. Livet krever stadig strøm av stoffer og energi gjennom veiene, og det krever at konsentrasjonene holdes langt fra likevekt (for eksempel høy [ATP]/[ADP]).
:::

**Oppgave 3.** Regn ut ΔG for hydrolyse av ATP ved 37 °C med [ATP] = 1,85 mM, [ADP] = 0,14 mM og [Pᵢ] = 1,0 mM. Hvorfor er verdien mer negativ enn ΔG°′?

::: {.callout-tip collapse="true"}
## Svarforslag
Q = (0,14·10⁻³ · 1,0·10⁻³)/(1,85·10⁻³) = 7,6·10⁻⁵. ΔG = −30,5 + 2,58 · ln(7,6·10⁻⁵) ≈ −30,5 − 24,5 ≈ −55 kJ/mol. Verdien er mer negativ fordi cellen holder ATP høyt og ADP og Pᵢ lavt: Q er mye mindre enn 1, så ATP ligger langt fra likevekt.
:::

**Oppgave 4.** Bruk fosfatstigen. Kan glukose-6-fosfat fosforylere ADP til ATP? Hva betyr svaret for reaksjonen glukose + ATP → glukose-6-fosfat + ADP?

::: {.callout-tip collapse="true"}
## Svarforslag
ΔG°′ for overføring = ΔG°′(hydrolyse G6P) − ΔG°′(hydrolyse ATP) = −13,8 − (−30,5) = +16,7 kJ/mol: ugunstig. Altså kan ikke G6P lage ATP. Reaksjonen glukose + ATP → G6P + ADP har derimot ΔG°′ = −30,5 − (−13,8) = −16,7 kJ/mol og er gunstig, så fosfatet går fra ATP til glukose.
:::

**Oppgave 5.** Forklar de fire grunnene til at ATP-hydrolyse er gunstig. Hvilke av de tre pilarene (elektrontetthet, vekselvirkninger, termodynamikk) bruker du for hver?

::: {.callout-tip collapse="true"}
## Svarforslag
Frastøtning mellom negative ladninger: elektrontetthet og vekselvirkninger (like ladninger). Entropi (ett til to molekyler): termodynamikk. Resonansstabilisering av fosfat: elektrontetthet (delokalisering). Bedre hydrering av ADP og Pᵢ: vekselvirkninger (ion–dipol og hydrogenbindinger med vann).
:::

**Oppgave 6.** Hvorfor gir fettsyrer mer energi per karbon enn glukose når de oksideres fullstendig?

::: {.callout-tip collapse="true"}
## Svarforslag
Fettsyrer er mer redusert: de har mange C–H-bindinger og få C–O-bindinger. Karbonene har mer elektrontetthet som kan «tas bort» ved oksidasjon. Glukose har allerede mange C–O-bindinger (er delvis oksidert), så det er mindre igjen å hente.
:::

**Oppgave 7.** Hva er forskjellen på NADH og NADPH når det gjelder bruk i cellen? Hvorfor trenger vi to ulike elektronbærere?

::: {.callout-tip collapse="true"}
## Svarforslag
NADH brukes mest i katabolismen, og leverer elektroner til elektrontransportkjeden for å lage ATP. NADPH brukes i anabolismen, og leverer elektroner til biosynteser (reduserende kraft). Ved å ha to separate pooler kan cellen regulere nedbrytning og oppbygging uavhengig av hverandre: NAD⁺ holdes mest oksidert, NADPH mest redusert.
:::

**Oppgave 8.** Hvorfor bryter cellen ned glukose i ti små trinn i stedet for ett stort?

::: {.callout-tip collapse="true"}
## Svarforslag
Energien kan fanges i små porsjoner (ATP og NADH) i stedet for å gå tapt som varme. Hvert trinn har dessuten sitt eget enzym, så cellen får mange punkter der veien kan reguleres.
:::
