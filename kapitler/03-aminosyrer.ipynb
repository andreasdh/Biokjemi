## Hvorfor lære om aminosyrer? {#sec-hvorfor-aminosyrer}

Proteiner gjør det aller meste i cellen. De er enzymer som katalyserer reaksjonene, de danner strukturer som cytoskjelettet og kollagenet i huden din, de frakter oksygen (hemoglobin), sender signaler (mange hormoner), og de forsvarer deg mot sykdom (antistoffer). Alle disse ulike jobbene gjøres av molekyler som er bygd av de samme tjue byggesteinene: **aminosyrene**.

Tenk på alfabetet. Med bare noen få bokstaver kan du skrive alt fra handleliste til roman, fordi det er rekkefølgen som gjør forskjellen. Slik er det også med proteiner. Sekvensen av aminosyrer bestemmer hvordan proteinet folder seg, og dermed hva det kan gjøre. Du husker det sentrale dogmet fra kapittel 1: DNA → RNA → protein. Det er aminosyresekvensen som koden i DNA til slutt bestemmer.

Hvor mange ulike «ord» kan du lage? Prøv selv:

{{< include ../_interaktiv/aa-kombinatorikk.qmd >}}

For å forstå proteiner må vi først kjenne byggesteinene godt. Kapittelet handler om tre ting, og de er bygd på de tre spørsmålene fra forrige kapittel:

::: {.nokkelpunkt}
**Struktur:** Hvordan er aminosyrer bygd opp, og hvordan kan vi gruppere dem? Her handler det om elektrontetthet: hvilke sidegrupper er upolare, polare eller ladde?

**Isomeri:** Aminosyrene finnes i to speilbildeformer, og naturen bruker bare den ene. Hvordan beskriver vi det?

**Syre-baseegenskaper:** Aminosyrene har grupper som kan ta opp og avgi protoner. Ladningen avhenger derfor av pH, og det avgjør hvilke vekselvirkninger de kan danne.
:::

## Generell struktur {#sec-aa-struktur}

Alle de 20 aminosyrene i proteiner har samme grunnstruktur. Et sentralt karbonatom, **α-karbonet** (Cα), binder fire ting:

1. en **aminogruppe** (–NH₂)
2. en **karboksylgruppe** (–COOH)
3. et **hydrogenatom**
4. en **sidegruppe** (R), som er forskjellig fra aminosyre til aminosyre

De tre første er felles for alle, og sidegruppa er det som gjør hver aminosyre spesiell. Utforsk delene her, og prøv å flytte protonet:

{{< include ../_interaktiv/aa-anatomi.qmd >}}

Aminosyrer har altså både en syre (COOH) og en base (NH₂) i samme molekyl. I vann, ved pH ≈ 7, overfører karboksylgruppa et proton til aminogruppa. Vi får et **zwitterion**: et molekyl med både en positiv og en negativ ladning, men med total ladning 0. Det er denne formen som dominerer i kroppen.

::: {.callout-note}
## Proteinogene aminosyrer
«Proteinogene» betyr «proteinbyggende»: aminosyrene som finnes i proteiner og er kodet direkte av DNA. Det er 20 av dem i standardkoden (pluss to spesielle, selenocystein og pyrrolysin, som vi ser på om litt).
:::

## De 20 aminosyrene {#sec-20-aminosyrer}

Du kan godt bruke tid på å bli kjent med alle 20, og det lønner seg. Heldigvis kan vi sortere dem etter hva sidegruppa gjør. Det er her elektrontetthet og vekselvirkninger fra kapittel 2 kommer inn igjen.

| Gruppe | Aminosyrer | Hva sidegruppa kan |
|:---|:---|:---|
| **Upolare** | Gly, Ala, Val, Leu, Ile, Phe, Trp, Met, Pro | Bare London-krefter. Hydrofobe, og samler seg i proteinets indre (den hydrofobe effekten) |
| **Polare** (uladde) | Ser, Thr, Cys, Asn, Gln, Tyr | Hydrogenbindinger (O–H, N–H, C=O). Cys kan også danne disulfidbroer |
| **Negative** | Asp, Glu | Ionebindinger (saltbroer) med positive grupper, og hydrogenbindinger |
| **Positive** | Lys, Arg, His | Ionebindinger med negative grupper, og hydrogenbindinger |

Merk at grensene ikke er skarpe. Tryptofan og tyrosin er store og delvis hydrofobe, selv om tyrosin har en OH-gruppe. Histidin kan være ladd eller uladd nær fysiologisk pH. Og glysin har bare et hydrogen som sidegruppe, så den passer egentlig dårlig i alle grupper.

Velg en aminosyre i oversikten under. Prøv å se på sidegruppa (den fargede flekken) og spør deg selv: *Hvor er elektronene? Hva kan denne gruppa vekselvirke med?* Du kan også endre pH og se hvordan ladningen endres.

## Test deg selv: aspartam og natriumglutamat
**Hvilke to aminosyrer består aspartam av? Hva med natriumglutamat?** Aspartam er asparaginsyre og fenylalanin, der fenylalanin er modifisert til en metylester (COOCH₃ i stedet for COOH). Natriumglutamat (MSG) er natriumsaltet av glutaminsyre.

**Hvilken reaksjonstype skjer når de to aminosyrene kobles sammen?** En kondensasjon: et vannmolekyl spaltes av, og de to aminosyrene bindes sammen med en peptidbinding.

**Hva brytes aspartam ned til i kroppen?** Til asparaginsyre, fenylalanin og metanol (fra esteren). Derfor må personer med fenylketonuri (PKU), som ikke tåler fenylalanin, unngå aspartam.
:::

## Fra aminosyre til peptid {#sec-peptid}

Når to aminosyrer kobles sammen, reagerer karboksylgruppa i den ene med aminogruppa i den andre. Det dannes en **peptidbinding** (en amidbinding), og et vannmolekyl frigjøres. Dette er en kondensasjonsreaksjon. Gjentar du det mange ganger, får du en kjede. Én ende har en fri aminogruppe (**N-terminalen**), og den andre har en fri karboksylgruppe (**C-terminalen**). Kjeden leses alltid fra N til C.

Det som går igjen langs hele kjeden (N–Cα–C), kalles **ryggraden**. Sidegruppene henger ut til sidene. Prøv selv: skriv navnet ditt med ettbokstavskoden, og se hvilken peptid det blir. (Vi bruker bokstavene A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, T, V, W og Y. Resten hoppes over.)

{{< include ../_interaktiv/aa-peptid.qmd >}}

Legg merke til at ladningen til peptidet endrer seg med pH. Den avhenger av de samme gruppene som vi ser på i slutten av kapittelet.

## Isomeri: speilbilder {#sec-isomeri}

### Hvorfor spiller det noen rolle?

Talidomid ble solgt som beroligende middel til gravide på slutten av 1950-tallet, og førte til at mer enn 10 000 barn ble født med alvorlige misdannelser. Molekylet er kiralt: det finnes i to speilbildeformer som ikke kan legges oppå hverandre, på samme måte som venstre og høyre hånd. Man har lenge fortalt at bare den ene formen var skadelig, men historien er mer komplisert (formene kan omdannes til hverandre i kroppen). Lærdommen står likevel: speilbilder av et molekyl kan oppføre seg helt ulikt i biologiske systemer, fordi kroppens molekyler selv er kirale.

Alle aminosyrer unntatt glysin er kirale, fordi Cα har fire ulike grupper. Glysin har to like (H og H) og er derfor ikke kiral.

### L og D: Fischer-projeksjon

Vi kaller de to formene **L** og **D**. Alle aminosyrer i proteiner er L-aminosyrer. D-aminosyrer finnes i naturen, men i svært liten grad (blant annet i noen bakterieveggers cellevegg).

For å skille dem tegner vi molekylet i en **Fischer-projeksjon**:

1. Sett karboksylgruppa øverst og sidegruppa nederst.
2. De vannrette bindingene peker ut mot deg, og de loddrette peker bort fra deg.
3. Se på hvilken side aminogruppa sitter: **venstre (laevus) = L, høyre (dexter) = D**.

Dra i molekylet nedenfor for å se det fra alle kanter, trykk «Fischer-visning» for å få det i riktig orientering, og bytt mellom L- og D-form:

```{ojs}
//| echo: false
lagKiralitet("LD")
```

::: {.callout-tip collapse="true"}
## Test deg selv: D eller L?
Fischer-projeksjonen av serin med COOH øverst, CH₂OH nederst, H₂N til venstre og H til høyre er **L-serin**. Hvis du bytter plass på H₂N og H, får du D-serin. Husk å plassere syregruppa øverst før du vurderer! Hvis molekylet er tegnet «på hodet», kan du rotere det 180° i planet. Det går bra, men aldri flip det ut av planet, for da får du speilbildet.
:::

### R og S: CIP-systemet

L og D passer for aminosyrer (og sukker), men er ikke et allmenngyldig system. Kjemikere bruker oftere **R/S-systemet** (Cahn–Ingold–Prelog), som fungerer for alle kirale molekyler:

1. Gi hver av de fire gruppene på det kirale karbonet en **prioritet** (1–4). Høyere atomnummer = høyere prioritet. Hvis to atomer er like, ser du på neste atom i rekken.
2. Snu molekylet slik at gruppa med **lavest prioritet** (for aminosyrer: H) peker bort fra deg.
3. Se på rekkefølgen 1 → 2 → 3. **Med klokka = R (rectus), mot klokka = S (sinister).**

For de aller fleste aminosyrene i proteiner blir prioriteten NH₂ (1) > COOH (2) > sidegruppe (3) > H (4), og alle blir da **S**. Men én aminosyre er et unntak. Finn ut hvilken ved å prøve alle fire:

```{ojs}
//| echo: false
lagKiralitet("RS")
```

::: {.callout-note}
## Cystein er R
Sidegruppa i cystein, CH₂SH, har svovel (nr. 16) på første plass, og svovel slår oksygen (nr. 8) i COOH. Prioritetene bytter plass, og selv om molekylet har samme romlige form som de andre L-aminosyrene, kalles det **R**. L/D sier noe om geometrien i forhold til serin eller glyseraldehyd, mens R/S følger prioritetsreglene. De to systemene er altså ikke det samme, og det er grunnen til at vi har begge.
:::

::: {.callout-tip collapse="true"}
## Test deg selv: R eller S?
Serin i Fischer-projeksjon (COOH øverst, H₂N til venstre, H til høyre, CH₂OH nederst) er L-serin. Prioritetene er NH₂ (1), COOH (2), CH₂OH (3), H (4). H peker ut mot deg, så vi må snu svaret: 1 → 2 → 3 går med klokka sett forfra, så det riktige svaret er mot klokka, altså **S**. Samme resultat som i figuren.
:::

## Syre-baseegenskaper {#sec-syre-base}

### pKₐ og isoelektrisk punkt

Ladningen til en aminosyre avhenger av pH, fordi gruppene kan ta opp og avgi protoner. Du kjenner sammenhengen fra forrige kapittel: en gruppe er halvparten protonert og halvparten deprotonert når pH = pKₐ (Henderson–Hasselbalch). Alanin har to slike grupper:

- **α-karboksylgruppa**, pKₐ₁ ≈ 2,3: er COOH ved lavere pH og COO⁻ ved høyere pH
- **α-aminogruppa**, pKₐ₂ ≈ 9,7: er NH₃⁺ ved lavere pH og NH₂ ved høyere pH

Det gir tre former: ladning +1 (sur løsning), 0 (zwitterion) og −1 (basisk løsning). Når vi titrerer med base, får vi en kurve med to «platåer», og midt mellom dem ligger **isoelektrisk punkt (pI)**: pH-verdien der aminosyren i gjennomsnitt er nøytral. For en aminosyre med to ioniserbare grupper er

$$\text{pI} = \frac{\text{p}K_{a1} + \text{p}K_{a2}}{2}$$

Aminosyrer med en ladd sidegruppe har tre pKₐ-verdier. Da tar vi gjennomsnittet av de to pKₐ-verdiene som ligger på hver side av den nøytrale formen. Velg en aminosyre og dra i pH-glidebryteren:

{{< include ../_interaktiv/aa-titrering.qmd >}}

Ved pI vandrer ikke aminosyren i et elektrisk felt. Over pI er den negativ og vandrer mot pluss-polen, og under pI er den positiv og vandrer mot minus-polen. Dette brukes i elektroforese og isoelektrisk fokusering til å skille proteiner fra hverandre.

Her er typiske pKₐ-verdier for ioniserbare grupper i proteiner. Verdiene for frie aminosyrer kan avvike litt:

| Gruppe | pKₐ (typisk) | Protonert → deprotonert |
|:---|:---:|:---|
| α-karboksyl (C-terminal) | 3,1 | COOH → COO⁻ |
| Asp og Glu (sidegruppe) | 4,1 | COOH → COO⁻ |
| His (sidegruppe) | 6,0 | imidazolium → imidazol |
| α-amino (N-terminal) | 8,0 | NH₃⁺ → NH₂ |
| Cys (sidegruppe) | 8,3 | SH → S⁻ |
| Lys (sidegruppe) | 10,8 | NH₃⁺ → NH₂ |
| Tyr (sidegruppe) | 10,9 | OH → O⁻ |
| Arg (sidegruppe) | 12,5 | guanidinium → guanidin |

Legg merke til at histidin har en pKₐ nær fysiologisk pH (7,4). Derfor kan histidin fungere som en syre eller base i mange enzymers aktive sete, og den fungerer som en liten buffer.

::: {.callout-tip}
## Oppgave: lysin
Lysin har pKₐ = 2,18 (α-COOH), 8,95 (α-NH₃⁺) og 10,5 (sidegruppe).

1. Tegn alle formene av lysin (ladde og uladde grupper).
2. Hvilken form dominerer ved pH 1, pH 7 og pH 12?
3. Regn ut pI. Hva forteller verdien deg?

Bruk titreringen over til å sjekke svaret ditt: velg Lys og se hvilken tilstand som er markert.
:::

::: {.callout-tip collapse="true"}
## Svarforslag
Lysin har fire former, med ladning +2 (pH < 2,18), +1, 0 og −1 (pH > 10,5). Ved pH 1 dominerer +2 (begge aminogrupper protonert, COOH), ved pH 7 +1 (COO⁻ og to NH₃⁺), og ved pH 12 −1 (COO⁻ og to NH₂). Den nøytrale formen ligger mellom pKₐ₂ og pKₐ₃, så pI = (8,95 + 10,5)/2 ≈ 9,7. At pI er så høy, forteller at lysin er en basisk aminosyre: den er positivt ladd i hele det fysiologiske pH-området.
:::

### Omgivelsene forskyver pKₐ

Verdiene i tabellen er typiske, men sidegruppene sitter ikke i rent vann. I et protein kan nabogrupper og omgivelser endre hvor lett en gruppe avgir et proton. Det er elektrontetthet og vekselvirkninger i praksis: ladninger tiltrekker og frastøter hverandre, og vann stabiliserer ladninger. Se på sidegruppa til glutaminsyre (Glu) i fire ulike omgivelser:

{{< include ../_interaktiv/aa-pka-miljo.qmd >}}

Regelen er: **alt som stabiliserer den ladde formen (COO⁻) senker pKₐ, og alt som destabiliserer den, øker pKₐ.** Nær en positiv gruppe blir COO⁻ stabilisert (saltbro), nær en negativ gruppe eller i et upolart miljø uten vann blir den destabilisert. For basiske grupper som lysin er det motsatt: det er den *positive* formen som må stabiliseres eller destabiliseres.

Det er dette som gjør at enzymer kan «stille inn» sine sure og basiske grupper til nøyaktig riktig pH-område.

## Oppsummering {#sec-aa-oppsummering}

::: {.nokkelpunkt}
**Struktur:** Alle aminosyrer har en aminogruppe, en karboksylgruppe, et H og en sidegruppe (R) på Cα. Ved pH 7 er de zwitterioner. De 20 proteinogene kan grupperes som upolare, polare, negative og positive, etter hva sidegruppa kan vekselvirke med.

**Isomeri:** Alle unntatt glysin er kirale. Proteiner består av L-aminosyrer. I R/S-systemet er nesten alle S, og cystein er R fordi S har høyere atomnummer enn O.

**Syre-baseegenskaper:** Hver ioniserbar gruppe har en pKₐ. Ladningen avhenger av pH, og pI er gjennomsnittet av de to pKₐ-verdiene rundt den nøytrale formen. Omgivelsene kan forskyve pKₐ.
:::

## Oppgaver {#sec-aa-oppgaver}

**Oppgave 1.** Tegn strukturen til en aminosyre med en positivt ladd sidegruppe, og forklar hvilken type vekselvirkning den kan danne med en negativt ladd sidegruppe.

::: {.callout-tip collapse="true"}
## Svarforslag
For eksempel lysin med –(CH₂)₄–NH₃⁺. Den danner en ionebinding (saltbro) med en negativt ladd gruppe som asparaginsyre (–CH₂–COO⁻). Tiltrekningen er elektrostatisk (Coulombs lov). I vann må saltbroen konkurrere med vannets ion–dipol-vekselvirkninger, så den er ofte nær overflaten eller i proteinets indre der det er lite vann.
:::

**Oppgave 2.** Hvorfor er leucin hydrofob, men serin ikke? Forklar med elektrontetthet og vekselvirkninger.

::: {.callout-tip collapse="true"}
## Svarforslag
Leucins sidegruppe er bare C og H. Forskjellen i elektronegativitet er så liten at C–H er upolar, så det finnes ingen delladninger og bare London-krefter er mulig. Serins CH₂–OH har en polar O–H-binding (δ⁻ på O og δ⁺ på H) og kan danne hydrogenbindinger med vann. Leucin må ordne vannmolekylene rundt seg (entropi), og samler seg derfor sammen med andre upolare grupper.
:::

**Oppgave 3.** Hva er forskjellen på en D-aminosyre og en L-aminosyre? Hvorfor er det ikke et problem at vi sier «alle proteiner består av L-aminosyrer»?

::: {.callout-tip collapse="true"}
## Svarforslag
De er speilbilder av hverandre: aminogruppa sitter til venstre (L) eller høyre (D) i Fischer-projeksjonen. Proteiner blir laget av ribosomet, som bare bruker L-aminosyrer, så de er alle L (unntatt glysin, som ikke har noen form).
:::

**Oppgave 4.** Aspartat (Asp) har pKₐ = 1,88 (α-COOH), 3,65 (sidegruppe) og 9,60 (α-NH₃⁺). Regn ut pI, og forklar hvorfor det er riktig å ta gjennomsnittet av de to første.

::: {.callout-tip collapse="true"}
## Svarforslag
pI = (1,88 + 3,65)/2 ≈ 2,8. Asp er positiv (+1) når alle tre gruppene er protonert, og nøytral når bare α-COOH er deprotonert. Den nøytrale formen finnes derfor mellom pKₐ₁ og pKₐ₂, så vi tar gjennomsnittet av dem. Fordi pI er så lav, er asparaginsyre en sur aminosyre.
:::

**Oppgave 5.** Et enzym har en glutaminsyre i det aktive setet som må være protonert (COOH) ved pH 7 for å virke. Hvordan kan omgivelsene gjøre det mulig?

::: {.callout-tip collapse="true"}
## Svarforslag
Vanligvis har Glu pKₐ ≈ 4,1 og er deprotonert (COO⁻) ved pH 7. Hvis gruppa sitter i et upolart miljø eller nær en negativ ladning, blir COO⁻ destabilisert, og pKₐ stiger. Er pKₐ høyere enn 7, vil gruppa være protonert og kunne virke som en syre i reaksjonen.
:::

**Oppgave 6.** Velg en aminosyre fra hver gruppe (upolar, polar, negativ, positiv) og forklar med de tre spørsmålene fra kapittel 2 hvorfor den havner der den gjør i et protein som folder seg i vann.
