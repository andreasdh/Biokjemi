Glukose er hjernens og de røde blodcellenes viktigste drivstoff, og utgangspunktet for stoffskiftet i nesten alle celler. Men glukose kan ikke bare «brennes»: cellen må ta den opp, gjøre den klar og tappe energien i små porsjoner. **Glykolysen** (fra gresk *glykys*, søt, og *lysis*, spalting) er veien som gjør dette. Den skjer i cytoplasma, trenger ikke oksygen og finnes i nesten alle levende celler.

Det er ti trinn, men du trenger ikke lære dem som en liste. Vi tar dem i tre akter, og for hvert trinn stiller vi tre spørsmål:

::: {.nokkelpunkt}
**Hva skjer?** Hvilken endring skjer i molekylet?

**Hvorfor?** Hva er hensikten, og hva driver det? (Tenk på elektrontetthet, vekselvirkninger og termodynamikk.)

**Hva skjer videre?** Hvor blir produktet brukt, og hva er koblet til andre prosesser?
:::

## Overblikk: glykolysen i tre akter {#sec-tre-akter}

Det er lettere å huske ti trinn når du ser mønsteret. Glykolysen er bygd opp som en **investering** etterfulgt av en **høsting**:

{{< include ../_figurer/glykolyse-akter.html >}}

1. **Akt 1: Investere (trinn 1–3).** Cellen bruker 2 ATP for å gjøre glukose klar. Det kan virke som et dårlig kjøp, men uten fosfatene kan ikke glukosen holdes inne eller spaltes.
2. **Akt 2: Dele (trinn 4–5).** Sukkeret med seks karbon deles i to like biter med tre karbon (G3P).
3. **Akt 3: Høste (trinn 6–10).** Hver av de to bitene gir 2 ATP og 1 NADH. Siden det er to biter, blir utbyttet 4 ATP og 2 NADH. Trekker vi fra investeringen, er netto **2 ATP** og **2 NADH**.

::: {.callout-note}
## Total reaksjon
$$\text{Glukose} + 2\,\text{NAD}^+ + 2\,\text{ADP} + 2\,\text{P}_i \longrightarrow 2\,\text{pyruvat} + 2\,\text{NADH} + 2\,\text{ATP} + 2\,\text{H}^+ + 2\,\text{H}_2\text{O}$$
:::

## Først inn i cellen: glukosetransportørene {#sec-glut}

Glukose er et stort, polart molekyl (mange OH-grupper) og kan ikke diffundere gjennom det upolare indre av membranen. Den trenger hjelp fra **glukosetransportører** (GLUT) i membranen. De virker som en kanal: de slipper glukose gjennom *nedover* konsentrasjonsgradienten, i begge retninger. De bruker ingen ATP.

Det finnes flere varianter, tilpasset ulike vev. Tallet **Kₘ** forteller hvilken glukosekonsentrasjon som gir halvparten av maksimalt opptak. Blodsukkeret ligger vanligvis på 4–6 mM.

| Transportør | Hvor | Kₘ | Kommentar |
|:---|:---|:---:|:---|
| GLUT1 | alle pattedyrceller | 1 mM | Grunnleggende opptak (mettet ved vanlig blodsukker) |
| GLUT2 | lever og β-celler i bukspyttkjertelen | 15–20 mM | Følger blodsukkeret: fungerer som glukosesensor og tar opp overskudd i leveren |
| GLUT3 | alle pattedyrceller (særlig nerveceller) | 1 mM | Grunnleggende opptak |
| GLUT4 | muskel- og fettceller | 5 mM | Styres av insulin; flere ved utholdenhetstrening |
| GLUT5 | tynntarm | 15 mM | Transporterer fruktose |

Når du har spist, stiger blodsukkeret. β-cellene i bukspyttkjertelen merker det (via GLUT2 og glykolysen) og skiller ut **insulin**. Insulin bindes til reseptorer på muskel- og fettceller og fører til at vesikler med GLUT4 smelter sammen med membranen. Da kan cellene ta opp mye mer glukose.

Men kanalen slipper glukose gjennom begge veier. Hvordan kan cellen da få *netto opptak*? Prøv selv, med og uten heksokinase, som er enzymet i første trinn av glykolysen:

{{< include ../_interaktiv/gl-glut.qmd >}}

Uten et enzym som bruker glukosen opp, vil konsentrasjonen inne bli lik den ute, og opptaket stopper. **Heksokinase** fosforylerer glukose til glukose-6-fosfat, som GLUT ikke kjenner igjen, og som dessuten er negativt ladd og ikke kommer gjennom membranen. Dermed holdes konsentrasjonen av fri glukose inne lav, og glukose fortsetter å strømme inn. Det er nøkkelen til første trinn.

## Trinn for trinn {#sec-trinn}

Nå følger vi glukose gjennom alle ti trinnene. Slik bruker du figuren:

- Velg en akt, eller trykk deg gjennom med «Neste».
- Strukturene vises som åpne kjeder, slik at du ser nøyaktig hva som endres. Den uthevede delen er det som skjer i trinnet. Glukose finnes i virkeligheten mest som ring, men kjeden er enklere å følge.
- Prøv å **gjette hvorfor** før du trykker for å se forklaringen. Hver gang er det én eller to enkle ideer bak.
- Nederst ser du løpende regnskap over ATP og NADH, og ΔG for trinnet i cellen.

{{< include ../_interaktiv/gl-stepper.qmd >}}

Andre sukker kan også gå inn i glykolysen på ulike steder, for eksempel fruktose. Når du har gått gjennom trinnene, kan du legge merke til tre mønstre som går igjen:

::: {.callout-tip}
## Tre mønstre
1. **Fosfater gir ladning og energi.** Fosfatgruppene holder mellomproduktene inne i cellen (negativ ladning), og gir dem evne til å gi videre fosfat.
2. **Kinaser bruker ATP, og noen lager ATP.** Trinn 1 og 3 bruker ATP som giver. I trinn 7 og 10 er det mellomproduktene (1,3-bisfosfoglyserat og PEP) som er bedre fosfatgivere enn ATP, så de lager ATP (fosfatstigen).
3. **Oksidasjon kobles til fosforylering.** I trinn 6 betaler energien fra oksidasjonen av aldehydet (NADH) for at det dannes et energirikt acylfosfat uten at ATP brukes.
:::

## Hvor sitter bremsene? {#sec-bremsene}

Ikke alle de ti trinnene er like. Om vi bare ser på ΔG°′ (standard), ser alt ut som en rolig strøm av både positive og negative verdier. Men i cellen, med de faktiske konsentrasjonene, ser bildet helt annerledes ut. ΔG er gitt av

$$\Delta G = \Delta G^{\circ\prime} + RT \ln Q$$

Utforsk det selv. Trykk på de ulike trinnene, bytt til kumulativ visning (slik forelesningens figur viser), og endre konsentrasjonene:

{{< include ../_interaktiv/gl-dgprofil.qmd >}}

Tre trinn skiller seg ut: **heksokinase (1)**, **PFK-1 (3)** og **pyruvatkinase (10)**. De har kraftig negativ ΔG i cellen, så de går i praksis bare én vei. De er **irreversible**. Alle de andre trinnene ligger nær likevekt (ΔG ≈ 0) og er **reversible**: enzymene virker begge veier, og retningen følger konsentrasjonene. Det gjelder selv aldolase, som på papiret ser umulig ut (ΔG°′ = +24), men som i cellen får nær null fordi produktene brukes opp med en gang.

### Hvorfor er det akkurat de tre som reguleres?

Tenk deg en elv. I de flate partiene flyter vannet bare dit nivåene tilsier, og du kan ikke styre mengden med en luke. Men der det er et stort fall, kan en luke bestemme hvor mye vann som passerer. Slik er det i glykolysen:

- Trinn med ΔG ≈ 0 følger bare det som skjer før og etter. Å skru på enzymaktiviteten der gjør nesten ingen forskjell for den totale flyten.
- Trinn med stor negativ ΔG er enveisventiler. Her **bestemmer enzymet hvor mye som slippes gjennom**, og her setter cellen inn bremsene.

Enzymene i de tre trinnene reguleres av signaler om cellens behov: **ATP** (høy energi), **AMP/ADP** (lav energi), **sitrat** (mange byggesteiner fra sitronsyresyklusen) og mellomproduktene selv. Se hvordan det fungerer:

{{< include ../_interaktiv/gl-regulering.qmd >}}

| Trinn | Enzym | Hemmes av | Aktiveres av |
|:---:|:---|:---|:---|
| 1 | Heksokinase | glukose-6-fosfat (produkthemming) | |
| 3 | PFK-1 | ATP, sitrat, H⁺ | AMP, ADP, fruktose-2,6-bisfosfat |
| 10 | Pyruvatkinase | ATP, alanin | fruktose-1,6-bisfosfat |

PFK-1 er hovedbryteren: det er der cellen *forplikter* seg til glykolysen. Fruktose-1,6-bisfosfat (produktet av PFK-1) aktiverer pyruvatkinase lenger ned (*fremovermating*), slik at flyten i starten og slutten henger sammen.

## Hva skjer videre? {#sec-videre}

### Hva skjer med NADH og pyruvat?

I trinn 6 ble NAD⁺ til NADH. Cellen har bare en liten pool av NAD⁺. Hvis NADH ikke kan levere elektronene sine videre, går NAD⁺ tom, trinn 6 stopper, og dermed hele glykolysen. Pyruvat og NADH må derfor ha en vei videre. Prøv deg frem:

{{< include ../_interaktiv/gl-skjebne.qmd >}}

- **Uten oksygen** (anaerob): pyruvat får elektronene fra NADH og blir til **laktat**, og NAD⁺ er tilbake. Det gir bare 2 ATP per glukose, men holder veien åpen. Slik gjør muskelcellene under sprint, og slik gjør røde blodceller hele tiden (de har ingen mitokondrier). Hos gjær blir pyruvat til etanol og CO₂ i stedet.
- **Med oksygen** (aerob): pyruvat går inn i mitokondriene og blir til **acetyl-CoA** (pyruvat + NAD⁺ + CoA → acetyl-CoA + CO₂ + NADH). Acetyl-CoA brennes helt ned i sitronsyresyklusen, og NADH leverer elektronene til elektrontransportkjeden. Da blir utbyttet mange ganger større. Det er temaet i de to neste kapitlene.

### Byggesteiner til andre veier

Glykolysen gir ikke bare energi. Mellomproduktene er også **utgangspunkt for andre molekyler**, og veiene kobles sammen:

| Mellomprodukt | Brukes også til |
|:---|:---|
| Glukose-6-fosfat | pentosefosfatveien (NADPH til biosyntese og ribose-5-fosfat til nukleotider), og lagring som glykogen |
| Dihydroksyacetonfosfat (DHAP) | glyserol-3-fosfat, ryggraden i triglyserider og fosfolipider |
| 3-fosfoglyserat | aminosyren serin (og derfra flere andre) |
| Pyruvat | alanin (aminosyre), oksaloacetat, acetyl-CoA (fettsyrer og sitronsyresyklusen), laktat |

## Den omvendte veien: glukoneogenese {#sec-glukoneogenese}

Noen celler (hjernen, røde blodceller) trenger stadig glukose, også når vi ikke har spist. Da lager **leveren** (og nyrene) ny glukose fra andre stoffer, først og fremst **laktat**, **aminosyrer** og **glyserol**. Prosessen kalles **glukoneogenese** («nydannelse av glukose»), og er nesten glykolysen baklengs.

Nesten. De syv reversible trinnene i midten (ΔG ≈ 0) kan brukes baklengs av de samme enzymene. Men de tre irreversible trinnene kan ikke bare snus: baklengs ville de hatt en kraftig *positiv* ΔG. Cellen må omgå dem med **andre enzymer** som koples til hydrolyse av ATP, GTP eller fosfat:

{{< include ../_interaktiv/gl-glukoneo.qmd >}}

Prisen er høy: for å lage én glukose fra to pyruvat trengs 4 ATP + 2 GTP + 2 NADH, mens glykolysen gir bare 2 ATP. Det er den prisen som gjør at *hele* veien oppover får en gunstig total ΔG.

Hvis glykolysen og glukoneogenesen gikk samtidig i samme celle, ville de bare brukt ATP og produsert varme (en «bortkastet syklus»). Derfor styres de **motsatt av de samme signalene**:

| Signal | PFK-1 (glykolyse) | Fruktose-1,6-bisfosfatase (glukoneogenese) |
|:---|:---:|:---:|
| AMP (lite energi) | aktiverer | hemmer |
| Fruktose-2,6-bisfosfat | aktiverer | hemmer |
| Sitrat (mange byggesteiner) | hemmer | aktiverer |

Tilsvarende hemmer ADP, og acetyl-CoA aktiverer, enzymet som tar første steg fra pyruvat mot glukose (pyruvat-karboksylase). Resultatet: avhengig av cellens status er det *enten* glykolyse *eller* glukoneogenese som er aktiv.

## Oppsummering {#sec-glykolyse-oppsummering}

::: {.nokkelpunkt}
**Glykolysen** gjør glukose (C₆) om til to pyruvat (C₃) i ti trinn: *investere* (2 ATP), *dele* og *høste* (4 ATP + 2 NADH). Netto: 2 ATP og 2 NADH.

**Fosforylering** (trinn 1 og 3) fanger glukose i cellen (ladning) og gjør den klar. ATP lages i trinn 7 og 10 ved fosforylering på substratnivå.

**Konsentrasjonsavhengig ΔG** (ΔG = ΔG°′ + RT ln Q) forklarer at de fleste trinn ligger nær likevekt, mens tre (heksokinase, PFK-1, pyruvatkinase) er irreversible kontrollpunkter.

**NAD⁺** må regenereres: med laktat uten oksygen, eller via mitokondriene med oksygen.

**Glukoneogenese** er nesten den omvendte veien, men må omgå de tre irreversible trinnene med andre enzymer og motsatt regulering.
:::

## Oppgaver {#sec-glykolyse-oppgaver}

**Oppgave 1.** Forklar hvorfor glukose-6-fosfat ikke forlater cellen, og hvorfor dette hjelper med å få glukose inn.

::: {.callout-tip collapse="true"}
## Svarforslag
Fosfatgruppen gir en negativ ladning, og GLUT kjenner bare igjen glukose, så glukose-6-fosfat kan ikke gå ut. Fosforyleringen holder også konsentrasjonen av *fri* glukose i cellen lav, så gradienten fra blodet inn i cellen opprettholdes og opptaket fortsetter.
:::

**Oppgave 2.** For aldolase er ΔG°′ = +23,8 kJ/mol. I en celle er [fruktose-1,6-bisfosfat] = 0,031 mM, [DHAP] = 0,14 mM og [G3P] = 0,019 mM. Regn ut ΔG ved 37 °C (RT = 2,58 kJ/mol). Hva forteller svaret?

::: {.callout-tip collapse="true"}
## Svarforslag
Q = [DHAP][G3P]/[FBP] = (0,14·10⁻³ · 0,019·10⁻³)/(0,031·10⁻³) ≈ 8,6·10⁻⁵. ΔG = 23,8 + 2,58 · ln(8,6·10⁻⁵) ≈ 23,8 − 24,1 ≈ −0,3 kJ/mol. Reaksjonen er altså omtrent i likevekt og kan gå i begge retninger, selv om ΔG°′ er svært positiv. Lave produktkonsentrasjoner (produktene brukes opp av neste trinn) gjør at Q er liten.
:::

**Oppgave 3.** Hvilke tre trinn i glykolysen er irreversible? Hvorfor er det nettopp disse som er kontrollpunkter?

::: {.callout-tip collapse="true"}
## Svarforslag
Heksokinase (1), PFK-1 (3) og pyruvatkinase (10). Alle har en stor negativ ΔG i cellen, så de går bare én vei. Bare slike trinn kan bestemme flyten gjennom veien. Trinn nær likevekt følger bare konsentrasjonene før og etter, så regulering av dem har liten effekt.
:::

**Oppgave 4.** En muskelcelle har høy [ATP] og høy [sitrat]. Hva skjer med PFK-1? Hva skjer med konsentrasjonen av fruktose-6-fosfat og glukose-6-fosfat, og hvilken effekt har det på heksokinase?

::: {.callout-tip collapse="true"}
## Svarforslag
ATP og sitrat hemmer PFK-1, så glykolysen bremses i trinn 3. Fruktose-6-fosfat hoper seg opp, og siden trinn 2 er reversibelt, gjør også glukose-6-fosfat det. Glukose-6-fosfat hemmer heksokinase (produkthemming), så bremsen forplanter seg bakover, og cellen tar opp mindre glukose.
:::

**Oppgave 5.** Regn ut netto utbytte av ATP og NADH per glukose, og forklar hvorfor vi må doble tallene i akt 3.

::: {.callout-tip collapse="true"}
## Svarforslag
Brukt: 2 ATP (trinn 1 og 3). Laget: 2 ATP × 2 (trinn 7 og 10, to G3P per glukose) = 4 ATP. Netto = 2 ATP. NADH: 1 × 2 = 2. Vi dobler fordi aldolase deler glukose i to tre-karbonmolekyler, og hvert av dem går gjennom trinn 6–10.
:::

**Oppgave 6.** Hvorfor lager musklene laktat under sprint? Hva skjer hvis cellen ikke kan lage laktat og ikke har oksygen?

::: {.callout-tip collapse="true"}
## Svarforslag
Under sprint rekker ikke oksygentilførselen å dekke behovet, så NADH kan ikke reoksideres i mitokondriene. Ved å overføre elektronene til pyruvat (som blir laktat) får cellen NAD⁺ tilbake. Uten laktat og oksygen går NAD⁺ tom, GAPDH (trinn 6) stopper, og hele glykolysen stopper.
:::

**Oppgave 7.** Trinn 6 danner en energirik fosfatbinding uten å bruke ATP. Hvor kommer energien fra?

::: {.callout-tip collapse="true"}
## Svarforslag
Fra oksidasjonen av aldehydet i G3P. Det frigjør nok energi til å betale for at fosfatet (fra Pᵢ) settes på. Oksidasjonen og fosforyleringen er koblet i samme enzym: elektronene går til NAD⁺, og fosfatet ender som 1,3-bisfosfoglyserat.
:::

**Oppgave 8.** Hvorfor kan ikke heksokinase, PFK-1 og pyruvatkinase bare «kjøres baklengs» i glukoneogenesen? Hvordan løser cellen det?

::: {.callout-tip collapse="true"}
## Svarforslag
Baklengs ville de ha en stor *positiv* ΔG (for eksempel ca. +34 kJ/mol for heksokinase), så reaksjonene går ikke av seg selv. Cellen bruker andre enzymer (glukose-6-fosfatase, fruktose-1,6-bisfosfatase, og pyruvat-karboksylase + PEP-karboksykinase) som kobles til hydrolyse av fosfat, ATP eller GTP, slik at hver omvei får en negativ ΔG.
:::
