## Eksempel: Gibbs fri energi

$$\Delta G = \Delta G^{\circ} + RT \ln Q$$

::: {.callout-note}
## Husk
Ved likevekt er $\Delta G = 0$.
:::

```python
#| label: fig-likevekt
#| fig-cap: "Andel produkt som funksjon av ΔG° ved 310 K."
import numpy as np
import matplotlib.pyplot as plt

R, T = 8.314e-3, 310  # kJ/(mol·K), K
dG0 = np.linspace(-30, 30, 200)
K = np.exp(-dG0 / (R * T))
produkt = K / (1 + K)

fig, ax = plt.subplots(figsize=(6, 3.4))
ax.plot(dG0, produkt, color="#3d7ea6", lw=2.5)
ax.axhline(0.5, color="#5aa889", ls="--", lw=1)
ax.set_xlabel("ΔG° (kJ/mol)")
ax.set_ylabel("Andel produkt")
for s in ("top", "right"):
    ax.spines[s].set_visible(False)
plt.show()
```

@fig-likevekt viser at små endringer i $\Delta G^{\circ}$ gir store utslag rundt null.
