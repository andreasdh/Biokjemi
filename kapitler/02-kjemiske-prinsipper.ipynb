Hvorfor holder de to trådene i DNA sammen? Hvorfor folder et protein seg til akkurat riktig form? Og hvorfor legger fettet i en membran seg i to pene lag, helt av seg selv?

Det overraskende er at du ikke trenger noen spesiell «biologisk» kjemi for å svare. Du trenger tre ideer fra kjemien du allerede kjenner. Når du møter et nytt molekyl eller en ny prosess, kan du stille de samme tre spørsmålene:

::: {.nokkelpunkt}
**1. Hvor er elektronene?** (elektrontetthet) Fordelingen av elektroner avgjør hvilke deler av et molekyl som er svakt positive, svakt negative, ladde eller upolare.

**2. Hva tiltrekker hva?** (molekylære vekselvirkninger) Ladninger og delladninger tiltrekker og frastøter hverandre. Resultatet er hydrogenbindinger, ionebindinger og London-krefter.

**3. Hva vil skje, og hvorfor?** (termodynamikk) Entalpi, entropi og fri energi forteller om en prosess skjer av seg selv, og hvor langt den går.
:::

Spørsmålene henger sammen som en kjede: elektronene bestemmer hvilke krefter som kan virke, og kreftene (sammen med uorden) bestemmer hva som skjer. Kapittelet er bygd opp rundt dem, og til slutt får du bruke alle tre på nye eksempler. Mye av dette har du møtt i generell og organisk kjemi. Her ser vi det i et nytt «setting»: i vann, ved 37 °C og i molekyler med tusenvis av atomer.

## Hvor er elektronene? {#sec-elektrontetthet}

Kjemiske bindinger handler om elektroner. I organisk kjemi har du sett at kovalente bindinger er delte elektronpar. I metan deler for eksempel karbon fire elektronpar med fire hydrogenatomer i fire σ-bindinger. Bindingene peker mot hjørnene i et tetraeder, med omtrent 109,5° mellom dem.

Kovalente bindinger er sterke. De utgjør skjelettet i alle biomolekyler, og de brytes ikke av tilfeldigheter. Men livet trenger mer enn et stivt skjelett. Molekylene må kunne finne hverandre, binde seg og slippe taket igjen, raskt og på riktig tidspunkt. De må altså kunne samhandle på en dynamisk og fleksibel måte. Det gjør de i vann, og for å forstå hvordan, må vi først se nærmere på hvordan elektronene er fordelt.

### Elektronegativitet

Atomer trekker ulikt hardt i elektronene i en binding. Dette kalles *elektronegativitet* (χ). Når to bundne atomer har ulik elektronegativitet, forskyves elektrontettheten mot det mest elektronegative atomet, som blir svakt negativt (δ⁻), mens det andre blir svakt positivt (δ⁺). Velg en binding, eller dra i glidebryteren:

{{< include ../_interaktiv/elektronegativitet.qmd >}}

Det finnes ikke noe skarpt skille mellom «delte» og «ikke-delte» elektroner. Bindingene ligger på et kontinuum: fra upolar kovalent (C–H), via polar kovalent (N–H og O–H) til ionisk (Na–Cl).

::: {.callout-tip}
## C–H er nesten upolar
Forskjellen mellom C (2,55) og H (2,20) er så liten at C–H regnes som upolar. Det er grunnen til at hydrokarboner, altså fett, oljer og mange aminosyresidekjeder, er «vannfryktende». Vi kommer tilbake til det i avsnittet om den hydrofobe effekten.
:::

### Vann: livets viktigste molekyl

Oksygen er mye mer elektronegativt enn hydrogen (3,44 mot 2,20). Elektrontettheten i vann forskyves derfor mot O, som blir δ⁻, mens H-atomene blir δ⁺. Siden molekylet er bøyd (omtrent 104,5°), summerer ikke de to bindingsdipolene seg til null. Hele molekylet har en positiv og en negativ side: vann er polart.

Det gir vann tre egenskaper som gjør det til et nesten perfekt medium for livet:

- Vann kan danne **hydrogenbindinger**, som ikke er kovalente og derfor kan brytes og dannes igjen hele tiden.
- Vann **løser polare og ladde stoffer**, fordi det kan vekselvirke med dem.
- Vann **blander seg ikke med upolare stoffer**. Det blir viktig om litt.

Molekyler i løsning beveger seg hele tiden (brownsk bevegelse). Om de finner hverandre og blir sammen, avhenger av hvilke vekselvirkninger de kan danne. Løselighet handler altså om vekselvirkninger.


Partialladningene er beregnet med en enkel modell og er veiledende. Poenget er mønsteret: de elektronegative atomene O og N er elektronrike (δ⁻), mens C og H som sitter på dem, er elektronfattige (δ⁺). Hydrokarboner har ingen utpregede ladninger.

## Hva tiltrekker hva? {#sec-vekselvirkninger}

Når du vet hvor ladningene sitter, kan du forutsi hvordan molekyler vekselvirker. I biokjemi spiller fire typer bindinger hovedrollene:

| Type | Hva er det? | Styrke (kJ/mol) | Eksempler |
|---|---|---|---|
| **Kovalent** | delte elektronpar | 150–500 | skjelettet i biomolekyler, peptidbindinger |
| **Ionebinding** | + tiltrekker − (elektronene er ikke delt) | 5–25 i vann | saltbroer, hydratiserte ioner |
| **Hydrogenbinding** | δ⁺ H mellom to elektronegative atomer (N, O) | 5–30 | vann, basepar i DNA, α-heliks og β-flak |
| **London-krefter** | midlertidige dipoler i upolare molekyler | 0,5–4 per kontakt | membraner, kjernen i proteiner, stabling av baser |

: De viktigste bindingstypene i biokjemi. Verdiene er omtrentlige og avhenger av omgivelsene. {.striped}

Legg merke til at de ikke-kovalente bindingene bare er noen få ganger sterkere enn varmebevegelsen i omgivelsene. Ved kroppstemperatur (37 °C) er RT ≈ 2,6 kJ/mol. Derfor brytes og dannes de hele tiden, og det er akkurat det livet trenger.

### Hydrogenbindinger

En hydrogenbinding oppstår når et δ⁺ hydrogen bundet til et elektronegativt atom (N eller O, kalt *donoren*) trekkes mot et frie elektronpar på et annet elektronegativt atom (*akseptoren*). Ingen elektroner deles, så det er ikke en kovalent binding.

Vann er det fremste eksempelet: hvert vannmolekyl kan danne opptil fire hydrogenbindinger, to som donor og to som akseptor. Hydrogenbindinger finner vi også *mellom* biomolekyler. I DNA parer A med T (to hydrogenbindinger) og G med C (tre). Det er plasseringen av donorer og akseptorer som gjør at akkurat disse parene passer. I proteiner binder C=O og N–H i peptidgruppene seg til hverandre og former α-heliks og β-flak. Ringene i figuren under viser hvor donorene (gule) og akseptorene (grønne) sitter. Du kan slå dem av og på.

```{ojs}
//| echo: false
lagMolekylVisning({ider: ["vannnettverk", "par_at", "par_gc", "par_peptid"],
  start: "vannnettverk", ph: false, steder: true})
```

## Konkurranse fra vann
I vann må en hydrogenbinding mellom to biomolekyler konkurrere med hydrogenbindingene til vann: C=O···H–N + H₂O···H₂O ⇌ C=O···H–OH + H–N···OH₂. Gevinsten ved å binde direkte er derfor liten for hver enkelt binding. Det er summen av mange som teller (se «Stabil, men fleksibel» lenger ned).
:::

### Ionebindinger

I en ionebinding er elektronene ikke delt, og bindingen er ren elektrostatikk. Vann kan likevel bryte opp ioneforbindelser, fordi det kan binde til både kationer og anioner: δ⁻ O vender mot Na⁺, og δ⁺ H vender mot Cl⁻. Dette kalles *hydratisering*. Også inne i proteiner finnes ladde sidekjeder, for eksempel Lys (+) og Asp (−). Når de møtes, kaller vi det en *saltbro*.


### London-krefter

Selv upolare molekyler kan tiltrekke hverandre. Elektronene i et molekyl er hele tiden i bevegelse, og i et øyeblikk kan fordelingen være litt skjev. Det gir en midlertidig dipol, som induserer en tilsvarende dipol i naboen. Resultatet er en svak tiltrekning. Den kalles London-krefter og er en variant av van der Waals-kreftene.

Hver kontakt er svak, men London-kreftene øker med kontaktflaten: jo større overflater som passer sammen, jo flere kontakter og jo sterkere samlet interaksjon. De dominerer mellom fettsyrehaler i membraner og i kjernen av proteiner. De finnes også mellom basene i DNA, som stables oppå hverandre som en myntrull. Faktisk bidrar stablingen mer til stabiliteten i dobbelthelikset enn hydrogenbindingene. Hydrogenbindingene sørger for at *riktig* base paret med *riktig* base.

### Hvem kan binde til hvem?

De funksjonelle gruppene avgjør hvilke bindinger som er mulige. Velg to grupper og se hva som kan skje mellom dem, og test deg selv på spørsmålet nederst.

{{< include ../_interaktiv/vekselvirkninger.qmd >}}

### Stabil, men fleksibel

Svake bindinger dannes og brytes lett. Det gir både stabilitet og fleksibilitet i biokjemiske prosesser. Hver sirkel under er en svak binding som brytes og dannes igjen ved 37 °C. Velg bindingstype og antall.

{{< include ../_interaktiv/svake-bindinger.qmd >}}

Én svak binding holder ingenting på plass. Men tjue holder sammen en struktur, samtidig som de enkelte bindingene fortsatt kan åpne seg. Slik får proteiner og DNA både stabilitet og fleksibilitet.

## Hva vil skje, og hvorfor? {#sec-termodynamikk}

Vi vet nå hvilke krefter som finnes. Men hva avgjør om en prosess faktisk skjer? Termodynamikk gir et enkelt og kraftfullt svar. For prosesser ved konstant temperatur og trykk, som i cellen, gjelder

$$\Delta G = \Delta H - T\,\Delta S$$

- **ΔG** (Gibbs fri energi) forteller om prosessen er spontan. ΔG < 0 betyr at den kan skje av seg selv, ΔG > 0 at den ikke kan det, og ΔG = 0 at systemet er i likevekt.
- **ΔH** (entalpi) er varmen som tas opp eller gis fra seg, og henger sammen med bindinger. *Bryter du bånd, må du tilføre energi (ΔH > 0). Danner du bånd, gir du fra deg varme (ΔH < 0).*
- **ΔS** (entropi) er et mål på uorden: hvor mange måter energi og partikler kan fordeles på. Mer frihet gir ΔS > 0. Temperaturen T avgjør hvor mye entropien teller.

Eksempler fra cellen:

- **Nedbryting av sukker** (glukose + O₂ → CO₂ + H₂O) har ΔG < 0 og er spontan.
- **Bygging av en DNA-kjede** fra frie nukleotider ordner mange molekyler til én kjede og har ΔG > 0. Den går ikke av seg selv.
- **Brunt fettvev** bruker fettnedbryting (fett + O₂ → CO₂ + H₂O, ΔH < 0) til å lage varme (termogenese), fordi det dannes sterke bindinger.
- **Smelting av fett** har ΔH > 0: varme brukes til å bryte de svake bindingene mellom lipidmolekylene. Men molekylene går fra en ordnet til en mer uordnet tilstand (ΔS > 0), og ved høy nok temperatur vinner entropien.

Prøv selv: skyv på ΔH, ΔS og temperaturen, eller velg et av eksemplene.

{{< include ../_interaktiv/termodynamikk.qmd >}}

Ser du mønsteret? Det er fire mulige kombinasjoner av fortegn:

| ΔH | ΔS | ΔG = ΔH − TΔS | Eksempel |
|---|---|---|---|
| − (varme avgis) | + (mer uorden) | alltid negativ: spontan | fettforbrenning |
| + (varme tas opp) | − (mindre uorden) | alltid positiv: aldri spontan | går bare hvis den kobles til en annen reaksjon |
| − (varme avgis) | − (mindre uorden) | negativ ved lav T | baseparing og folding ved lave temperaturer |
| + (varme tas opp) | + (mer uorden) | negativ ved høy T | fett som smelter, DNA som «smelter» |

: Fortegnene til ΔH og ΔS og hva de betyr for spontaniteten.

::: {.callout-important}
## Spontan betyr ikke rask
En sukkerbit og oksygen i luften reagerer spontant (ΔG < 0), men en sukkerbit i kjøkkenskuffen brenner ikke av seg selv. ΔG sier at noe *kan* skje, ikke *hvor fort*. Hastigheten avgjøres av kinetikk, og det er der enzymene kommer inn i senere kapitler.
:::

::: {.callout-tip}
## Hvordan kan cellen likevel bygge DNA?
En reaksjon med ΔG > 0 kan likevel skje hvis den kobles til en reaksjon med tilstrekkelig negativ ΔG. Det vanligste er hydrolyse av ATP (ΔG°′ ≈ −30 kJ/mol). Fri energi er additiv, så summen blir negativ. Cellens energimolekyler brukes opp og bygges opp igjen hele tiden. Mer om det i metabolismedelen.
:::

## Alle tre sammen: den hydrofobe effekten {#sec-hydrofob}

Du vet at olje og vann ikke blander seg. Men hvorfor egentlig? Det er fristende å si at oljemolekylene «liker» hverandre bedre enn vann. Det stemmer bare delvis. Se på hva de tre spørsmålene sier:

- **Hvor er elektronene?** I oljen (hydrokarboner) er C–H nesten upolar. Det finnes ingen δ⁺ og δ⁻ som kan lage hydrogenbindinger med vann.
- **Hva tiltrekker hva?** Vannmolekylene vil danne så mange hydrogenbindinger som mulig med hverandre. Rundt et upolart molekyl må de ordne seg på en spesiell måte for å klare det, og det begrenser hvor fritt de kan bevege seg.
- **Hva vil skje, og hvorfor?** Når de upolare molekylene samles, blir den totale overflaten mot vann mindre. Vannmolekyler frigjøres fra det ordnede skallet rundt dem, og entropien i vannet øker (ΔS > 0). Det gjør ΔG negativ.

Den hydrofobe effekten er derfor *entropidrevet*: samlingen av upolare molekyler skyldes i stor grad vannets frihet, mer enn en sterk tiltrekning mellom de upolare molekylene.

### Vokteroppgaven

Her er en oppgave du kan tenke deg til. Du har 4 kriminelle som skal i fengsel, og 20 voktere. Du skal passe på de kriminelle på en måte som gjør at flest mulig voktere kan svirre rundt og gjøre andre ting. De kriminelle er upolare molekyler, og vokterne er vannmolekylene som må ordne seg rundt dem. Hvordan bør du plassere de kriminelle?

{{< include ../_interaktiv/vokter.qmd >}}

Når de kriminelle samles, trengs det færre voktere. De som slipper fri, kan bevege seg mer enn før. På samme måte får vann mer frihet, altså høyere entropi, når upolare molekyler samles.

### Hvor finner vi den hydrofobe effekten?

- **Membraner:** Fosfolipider har et polart hode og upolare haler. I vann samler halene seg inni et dobbeltlag, og hodene vender mot vann. Slik dannes membraner av seg selv.
- **Proteinfolding:** Upolare sidekjeder (som Leu, Val og Phe) gjemmer seg i kjernen av proteinet, mens polare og ladde sidekjeder vender mot vann.
- **Såpe og vaskemidler** har samme oppbygning som fosfolipider og pakker inn fett i små kuler (miceller).
- **Binding av små molekyler:** Mange legemidler og hormoner bindes i upolare lommer i proteiner.

### Se et protein «kollapse»

En enkel modell av proteinfolding er HP-modellen: en kjede av hydrofobe (H, gule) og polare (P, grønne) perler på et gitter. For hver kontakt mellom to H-perler som ikke er naboer i kjeden, synker energien med én enhet. Det representerer at vann presses ut når upolare grupper samles. Simuleringen under flytter kjeden tilfeldig (Monte Carlo-simulering), med mer tilfeldighet ved høyere temperatur. Prøv ulik temperatur og ulike kjeder, eller trykk på «Kjøl sakte ned».

{{< include ../_interaktiv/hp-folding.qmd >}}

Ved lav temperatur vinner energien: kjeden kollapser til en kompakt form med en hydrofob kjerne. Ved høy temperatur vinner entropien: kjeden har uendelig mange måter å ligge på, og strukturen «smelter». Det samme skjer med virkelige proteiner når du koker et egg.

## pH og buffere: når ladningen avhenger av omgivelsene {#sec-ph}

Mange funksjonelle grupper kan ta opp eller gi fra seg protoner. Da endrer elektrontettheten seg, og dermed ladningen. Og ladningen avgjør hvilke vekselvirkninger gruppen kan danne. Du så det i molekylvisningen over: karboksylgruppen er nøytral (–COOH) ved lav pH og negativ (–COO⁻) ved høy pH, mens aminogruppen er positiv (–NH₃⁺) ved lav pH og nøytral (–NH₂) ved høy pH.

For en svak syre HA ⇌ H⁺ + A⁻ forteller Henderson–Hasselbalch-ligningen hvor stor andel som er på hver form:

$$\mathrm{pH} = \mathrm{p}K_a + \log\frac{[\mathrm{A^-}]}{[\mathrm{HA}]}$$

Når pH = pKₐ, er halvparten på hver form. Én pH-enhet over pKₐ er 90 % deprotonert, og én enhet under er 90 % protonert.

Derfor er pH så viktig. Blod har pH 7,4, og avvik på mer enn omtrent 0,5 enheter i begge retninger er livstruende. Bindinger forstyrres, proteiner foldes ikke riktig, enzymer virker ikke, og saltbalansen i cellene endres. Kroppen holder pH stabil med **buffere**: blandinger av en svak syre og dens konjugerte base, som motvirker endringer i pH.

Velg en syre og tilsett base. Merk deg hvor pH nesten ikke endrer seg, og sammenlikn med rent vann (stiplet linje).

{{< include ../_interaktiv/buffer.qmd >}}

::: {.callout-note}
## Hvor liker cellen lav pH?
I lysosomene er pH omtrent 4,5–5, mens cytosol har pH omtrent 7,2. Fordøyelsesenzymene i lysosomene virker best ved lav pH. Hvis noen lekker ut i cytosol, blir de mye mindre aktive, og det beskytter cellen. Her er pH-forskjellen altså både en forutsetning for funksjonen og en sikkerhetsmekanisme.
:::

## Tre spørsmål, mange fenomener {#sec-linser}

Du har nå brukt de samme tre spørsmålene på vann, DNA-baser, membraner, proteiner, ioner, pH og energi. Her er en oversikt over åtte eksempler. Velg ett og les hva hvert spørsmål sier om det. Slå på øvemodus for å prøve selv først, og gå deretter gjennom forklaringene én for én.

{{< include ../_interaktiv/linser.qmd >}}

Legg merke til at forklaringene ikke er tre separate historier, men én. Elektrontettheten avgjør hvilke vekselvirkninger som er mulige, og vekselvirkningene sammen med entropien avgjør hva som skjer. Når du møter et molekyl eller en prosess du aldri har sett før, har du alltid de tre spørsmålene å begynne med.

## Oppsummering {#sec-oppsummering-kap2}

- **Kovalente bindinger** er skjelettet i biomolekylene. De **ikke-kovalente** (hydrogenbindinger, ionebindinger og London-krefter) er svake, og derfor dynamiske: de gir både stabilitet og fleksibilitet.
- **Elektronegativitet** gir polare bindinger med δ⁺ og δ⁻. Vann er polart og kan lage hydrogenbindinger, mens C–H er upolar.
- **Hydrogenbindinger** gjør vann til et dynamisk medium og gir spesifikk gjenkjenning mellom baser og mellom peptidgrupper.
- **ΔG = ΔH − TΔS** avgjør spontaniteten. ΔG < 0 betyr spontan, men spontan betyr ikke rask.
- **Den hydrofobe effekten** er entropidrevet: når upolare molekyler samles, frigjøres vann, og entropien øker. Det forklarer membraner og proteinfolding.
- **pH** endrer ladning og dermed vekselvirkninger. **Buffere** holder pH stabil innenfor ca. ±1 enhet fra pKₐ.
- Tre spørsmål (**elektrontetthet, vekselvirkninger og termodynamikk**) går igjen og kan brukes på alt du møter.

## Oppgaver {#sec-oppgaver-kap2}

Prøv å svare med egne ord før du åpner svarforslaget.

**Oppgave 1.** Hvilke av de funksjonelle gruppene (hydrokarbon, aromatisk ring, hydroksyl, karbonyl, amid, karboksylat, ammonium, fosfat, tiol) kan *ikke* danne polare eller ioniske interaksjoner? Begrunn svaret med elektrontettheten.

::: {.callout-tip collapse="true"}
## Svarforslag
Hydrokarbon og aromatisk ring. De har nesten ingen delladninger (liten Δχ mellom C og H), så de verken kan gi eller ta imot hydrogenbindinger, og de er ikke ladd. Tiolen er et grensetilfelle fordi S–H er bare svakt polar.
:::

**Oppgave 2.** Tegn to vannmolekyler med en hydrogenbinding mellom dem. Merk av δ⁺, δ⁻, donor og akseptor.

**Oppgave 3.** Gi eksempler på molekyler i kroppen som *ikke* er blant de fire hovedtypene (proteiner, nukleinsyrer, lipider og karbohydrater). Hva slags vekselvirkninger kan de danne?

::: {.callout-tip collapse="true"}
## Svarforslag
Vann, ioner (Na⁺, K⁺, Ca²⁺, Cl⁻), mange vitaminer og små mellomprodukter i metabolismen. Også ATP, som er et enkelt nukleotid og ikke en nukleinsyre. Ioner danner ion–dipol-vekselvirkninger med vann og saltbroer, vann danner hydrogenbindinger, og ATP har ladde fosfatgrupper.
:::

**Oppgave 4.** Hvorfor er ikke én hydrogenbinding nok til å holde en struktur på plass, mens tjue kan være det? Bruk RT ≈ 2,6 kJ/mol i svaret.

::: {.callout-tip collapse="true"}
## Svarforslag
En enkelt hydrogenbinding på ca. 10 kJ/mol er bare noen ganger sterkere enn varmebevegelsen, så den brytes og dannes hele tiden. Med mange bindinger er det svært lite sannsynlig at alle er brutt samtidig, og strukturen er stabil. Hver enkelt binding kan likevel åpne seg, og det gir fleksibilitet.
:::

**Oppgave 5.** Fett smelter når det varmes opp (ΔH > 0). Forklar hvorfor det er spontant ved høy temperatur, og hvordan du kan finne smeltetemperaturen.

::: {.callout-tip collapse="true"}
## Svarforslag
ΔG = ΔH − TΔS. Siden ΔS > 0 (væske er mer uordnet enn fast stoff), blir −TΔS mer negativ når T øker. Når T·ΔS blir større enn ΔH, er ΔG < 0. Smeltetemperaturen er der ΔG = 0, altså T = ΔH/ΔS.
:::

**Oppgave 6.** I brunt fettvev brukes fettnedbryting til å lage varme. Hva er fortegnet til ΔH? Hva skjer med bindingene?

::: {.callout-tip collapse="true"}
## Svarforslag
ΔH < 0: varme avgis. Bindingene i fettmolekylene og O₂ brytes, men det dannes sterkere bindinger i CO₂ og H₂O. Netto frigjøres energi, og i brunt fettvev blir den til varme i stedet for å lagres i ATP.
:::

**Oppgave 7.** Forklar vokteroppgaven med egne ord. Hva tilsvarer de kriminelle, vokterne og de frie vokterne i vann? Hvorfor samler upolare molekyler seg, selv om de ikke tiltrekker hverandre særlig sterkt?

::: {.callout-tip collapse="true"}
## Svarforslag
Kriminelle er upolare molekyler. Vokterne er vannmolekyler som må ordne seg rundt dem. Når de upolare molekylene samles, blir overflaten mot vann mindre, og færre vannmolekyler må være ordnet. De som slipper fri, kan bevege seg mer, så entropien i vannet øker. ΔG blir negativ fordi −TΔS blir mer negativ, ikke fordi London-kreftene er sterke.
:::

**Oppgave 8.** Hvor i cellen liker vi å ha lavere pH, og hvorfor? Gi minst ett eksempel.

::: {.callout-tip collapse="true"}
## Svarforslag
I lysosomene (pH ca. 4,5–5). Fordøyelsesenzymene der virker best ved lav pH, og hvis de lekker ut i cytosol (pH ca. 7,2), blir de mye mindre aktive. Et annet eksempel er mellomrommet mellom de to membranene i mitokondriet, der en høy protonkonsentrasjon brukes til å lage ATP.
:::

**Oppgave 9.** Du skal lage en buffer ved pH 6,9. Hvilken av syrene i bufferfiguren er best egnet, og hvorfor? Hva med eddiksyre/acetat (pKₐ = 4,76) ved pH 7,4?

::: {.callout-tip collapse="true"}
## Svarforslag
Fosfat (pKₐ = 6,86) er best, fordi pH ligger innenfor pKₐ ± 1. Eddiksyre/acetat ved pH 7,4 er nesten bare deprotonert (acetat), siden pH er langt over pKₐ, og lite igjen av syreformen til å motvirke tilsatt base.
:::

**Oppgave 10.** Velg et eksempel som ikke står i kapittelet (for eksempel hvorfor sukker løses i vann, eller hvorfor sterke syrer ødelegger proteiner), og forklar det med de tre spørsmålene: elektrontetthet, vekselvirkninger og termodynamikk.
